```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
df = pd.read_csv("Organization_ECommerce_Data_Analytics_Assignment.csv")
print(df)
```

```python
print(f"Dataset Shape: {df.shape[0]} records and {df.shape[1]} columns.\n")
print("--- Data Types ---")
print(df.dtypes)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
print(f"\nDate Range: From {df['Date'].min().date()} to {df['Date'].max().date()}")
print("\n--- Unique Counts ---")
print(f"Customers: {df['Customer_ID'].nunique()}")
print(f"Products: {df['Product_ID'].nunique()}")
print(f"Categories: {df['Category'].nunique()}")
print(f"Regions: {df['Region'].nunique()}")
print(f"Customer Segments: {df['Customer_Segment'].unique().tolist()}")
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
print(f"Numerical Variables : {numerical_cols}")
```

```python
print("--- Missing Values ---")
print(df.isna().sum()[df.isna().sum()>0])

print("\n--- Duplicate Records ---")
duplicate_count = df.duplicated().sum()
print(f"Total duplicate records: {duplicate_count}")

print("\n--- Inconsistent Values Check ---")
print("Payment Methods before cleaning:", df['Payment_Method'].unique())

df['Payment_Method'] = df['Payment_Method'].str.title()
print("Payment Methods after cleaning:", df['Payment_Method'].unique())


print("\n--- Suspicious Values Check ---")
print(f"Transactions with Quantity <= 0: {(df['Quantity'] <= 0).sum()}")
print(f"Transactions with Delivery Days > 14: {(df['Delivery_Days'] > 14).sum()}")
print(f"Transactions with negative Profit: {(df['Profit'] < 0).sum()}")
```

```python
total_gross_sales = df['Gross_Sales'].sum()
total_net_sales = df['Net_Sales'].sum()
avg_transaction_value = df['Net_Sales'].mean()

print(f"Total Gross Sales: ₹{total_gross_sales:,.2f}")
print(f"Total Net Sales: ₹{total_net_sales:,.2f}")
print(f"Average Transaction Value: ₹{avg_transaction_value:,.2f}")

total_profit = df['Profit'].sum()
avg_profit = df['Profit'].mean()
profit_margin = (total_profit / total_net_sales) * 100

print(f"\nTotal Profit: ₹{total_profit:,.2f}")
print(f"Average Profit: ₹{avg_profit:,.2f}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")

avg_customer_spend = total_net_sales / df['Customer_ID'].nunique()
transactions_per_customer = len(df) / df['Customer_ID'].nunique()

print(f"\nNumber of customers : {df['Customer_ID'].nunique()}")
print(f"Average Spend per Customer: ₹{avg_customer_spend:,.2f}")
print(f"Average Transactions per Customer: {transactions_per_customer:.2f}")

top_products_sales = df.groupby('Product')['Net_Sales'].sum().sort_values(ascending=False).head(3)
low_products_profit = df.groupby('Product')['Profit'].sum().sort_values(ascending=True).head(3)

print("\n--- Top 3 Products by Sales ---")
print(top_products_sales)
print("\n--- Bottom 3 Products by Profit ---")
print(low_products_profit)
```

```python
category_eda = df.groupby('Category').agg(
    Total_Sales=('Net_Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Order_Value=('Net_Sales', 'mean')
).reset_index()
print("--- Category EDA ---")
display(category_eda)

region_eda = df.groupby('Region').agg(
    Total_Sales=('Net_Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Transactions=('Transaction_ID', 'count')
).reset_index()
print("\n--- Region EDA ---")
display(region_eda)

segment_eda = df.groupby('Customer_Segment').agg(
    Total_Spend=('Net_Sales', 'sum'),
    Total_Orders=('Transaction_ID', 'count'),
    Profit_Contribution=('Profit', 'sum')
).reset_index()
print("\n--- Customer Segment EDA ---")
display(segment_eda)

channel_eda = df.groupby('Sales_Channel').agg(
    Total_Sales=('Net_Sales', 'sum'),
    Total_Transactions=('Transaction_ID', 'count')
).reset_index()
print("\n--- Sales Channel EDA ---")
display(channel_eda)
```

```python
num_cols = ['Quantity', 'Unit_Price', 'Discount_Percent', 'Customer_Rating', 'Delivery_Days', 'Net_Sales', 'Profit']
corr_matrix = df[num_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()

print("\nKey Correlation Insights to Note:")
print("1. Correlation between Discount_Percent and Profit:", round(corr_matrix.loc['Discount_Percent', 'Profit'], 2))
print("2. Correlation between Delivery_Days and Customer_Rating:", round(corr_matrix.loc['Delivery_Days', 'Customer_Rating'], 2))
```

```python
def find_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return outliers

delivery_outliers = find_outliers_iqr(df, 'Delivery_Days')
profit_outliers = find_outliers_iqr(df, 'Profit')

print(f"Number of Delivery Days outliers: {len(delivery_outliers)}")
print(f"Number of Profit outliers: {len(profit_outliers)}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(y=df['Delivery_Days'], ax=axes[0], color='skyblue').set_title('Delivery Days Outliers')
sns.boxplot(y=df['Profit'], ax=axes[1], color='salmon').set_title('Profit Outliers')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 10))

plt.subplot(2, 2, 1)
sales_trend = df.groupby(df['Date'].dt.to_period('M'))['Net_Sales'].sum()
sales_trend.plot(kind='line', marker='o', color='green')
plt.title('Monthly Net Sales Trend')
plt.xlabel('Month')
plt.ylabel('Net Sales (₹)')

plt.subplot(2, 2, 2)
sns.barplot(x='Region', y='Total_Sales', data=region_eda, palette='viridis')
plt.title('Total Sales by Region')
plt.ylabel('Net Sales (₹)')

plt.subplot(2, 2, 3)
sns.barplot(x='Category', y='Total_Profit', data=category_eda, palette='magma')
plt.title('Total Profit by Category')
plt.ylabel('Profit (₹)')

plt.subplot(2, 2, 4)
sns.scatterplot(x='Discount_Percent', y='Profit', hue='Category', data=df, alpha=0.7)
plt.title('Discount Percent vs. Profit')
plt.axhline(0, color='red', linestyle='--')

plt.tight_layout()
plt.show()
```

```python
unprofitable_high_sales = df[df['Profit'] < 0].groupby('Product').agg(
    Total_Net_Sales=('Net_Sales', 'sum'),
    Total_Loss=('Profit', 'sum')
).sort_values(by='Total_Loss').head()
print("--- High Sales, Negative Profit Products ---")
display(unprofitable_high_sales)

aov_channel = df.groupby('Sales_Channel')['Net_Sales'].mean().sort_values(ascending=False)
print("\n--- Average Order Value by Channel ---")
display(aov_channel)

segment_discount = df.groupby('Customer_Segment')['Discount_Percent'].mean().sort_values(ascending=False)
print("\n--- Average Discount Percent by Customer Segment ---")
display(segment_discount)

fast_delivery_rating = df[df['Delivery_Days'] <= 7]['Customer_Rating'].mean()
slow_delivery_rating = df[df['Delivery_Days'] > 7]['Customer_Rating'].mean()
print(f"\n--- Delivery Speed vs. Ratings ---")
print(f"Avg Rating (<= 7 days): {fast_delivery_rating:.2f}")
print(f"Avg Rating (> 7 days): {slow_delivery_rating:.2f}")

payment_segment_ct = pd.crosstab(df['Customer_Segment'], df['Payment_Method'], normalize='index') * 100
print("\n--- Payment Method Distribution by Segment (%) ---")
display(payment_segment_ct.round(2))
```
